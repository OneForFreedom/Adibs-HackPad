# Part 2: Working with Colors & Text

Here's how to create gradients with pixel manipulation:

````python
# Here's a colorful gradient example
width, height = 300, 300
img = Image.new('RGB', (width, height))

# This creates a simple gradient - try to understand how!
for x in range(width):
    for y in range(height):
        r = int((x / width) * 255)
        g = int((y / height) * 255)
        b = 100
        img.putpixel((x, y), (r, g, b))

plt.imshow(img)
plt.axis('off')
plt.show()
````

## Challenge 2: Custom Gradient 🎯

**Your task:** Modify the gradient code to create a different color pattern. Try changing how `r`, `g`, and `b` are calculated.

````python
from PIL import Image
import matplotlib.pyplot as plt
import math


width, height = 400, 400
img = Image.new('RGB', (width, height))


for x in range(width):
    for y in range(height):
        r = int((math.sin(x / 30) + 1) * 5)
        g = int((math.cos(y / 30) + 1) * 150)
        b = int((math.sin((x + y) / 50) + 1) * 533)
        img.putpixel((x, y), (r, g, b))


plt.imshow(img)
plt.axis('off')
plt.show()
````

````python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt


x = np.linspace(-2*np.pi, 2*np.pi, 400)
y = np.linspace(-2*np.pi, 2*np.pi, 400)
X, Y = np.meshgrid(x, y)


Z = np.sin(X**2 + Y**2) + np.cos(X - Y)


Z_norm = ((Z - np.min(Z)) / np.ptp(Z) * 255).astype(np.uint8)


pattern_img = Image.fromarray(np.stack([Z_norm]*3, axis=-1))
pattern_img = pattern_img.resize((800, 800))


plt.imshow(pattern_img)
plt.axis('off')
plt.show()
````
